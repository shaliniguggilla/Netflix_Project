# Netflix Data Analysis Project

```python
import pandas as pd
```

```python
df = pd.read_csv("netflix_titles.csv")
df
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df["director"] = df["director"].fillna("Unknown")
df.head()
```

```python
df["cast"] = df["cast"].fillna("Unknown")
df["country"] = df["country"].fillna("Unknown")
df["rating"] = df["rating"].fillna("Not Rated")
df["duration"] = df["duration"].fillna("Unknown")
df.head()
```

```python
import warnings
warnings.filterwarnings("ignore")
df["date_added"] = df["date_added"].fillna(method = 'ffill')
df["date_added"] = pd.to_datetime(df["date_added"])
df.head()
```

```python
df.isnull().sum()
```

```python
df['type'].value_counts()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.countplot(x = 'type', data = df)
plt.title("Movies vs TV Shows on Netflix")
plt.show()
```

```python
top_countries = df['country'].value_counts().head(10)

top_countries.plot(kind  = 'bar')
plt.title("Top 10 countries Producing Netflix content")
plt.xlabel("Country")
plt.ylabel("Number of titles")
plt.show()
```

```python
Release_Year = df['release_year'].value_counts().head(20)
Release_Year.plot(kind = 'bar')
plt.title("Year Wise Releasing Movies List")
plt.xlabel("Release Year")
plt.ylabel("Title")
plt.show()
```

```python
df['release_year'].value_counts().sort_index().plot()

plt.title("Netflix Content Growth by Year")
plt.xlabel("Year")
plt.ylabel("Number of Titles")
plt.show()
```

```python
top_genres = df['listed_in'].value_counts().head(10)

top_genres.plot(kind = 'barh')
plt.title("Top Genres on Netflix")
plt.show()
```

```python
df['rating'].value_counts().plot(kind = 'bar')

plt.title("Distribution Of Ratings")
plt.xlabel("Rating")
plt.ylabel("Count")
plt.show()
```

```python
movies  = df[df['type'] == 'Movie']

movies['release_year'].value_counts().sort_index().plot()

plt.title("Movies Released Per Year")
plt.show()
```
